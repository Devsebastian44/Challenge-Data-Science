# Extracción de datos

## Importación de librerías y lectura de bases de datos



```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

# 1. Análisis de facturación



## **Ingreso Total por cada tienda**

### Descripcion:
En este primer análisis, debemos calcular el ingreso total de cada tienda. Sumando los valores de la columna Precio de cada conjunto de datos de la tienda para estimar los ingresos.

```python
suma_total_tienda1 = tienda['Precio'].sum()
suma_total_tienda2 = tienda2['Precio'].sum()
suma_total_tienda3 = tienda3['Precio'].sum()
suma_total_tienda4 = tienda4['Precio'].sum()
```

```python
print(f" El valor total de la tienda es: {suma_total_tienda1}")
print(f" El valor total de la tienda 2 es: {suma_total_tienda2}")
print(f" El valor total de la tienda 3 es: {suma_total_tienda3}")
print(f" El valor total de la tienda 4 es: {suma_total_tienda4}")
```

#### La tienda con mayor ingresos fue la **Tienda 1**, con un ingreso de **$ 1.150.880.400**

## **Visualizacion Grafica**

```python
suma_total_tienda1 = 1150880400.0
suma_total_tienda2 = 1116343500.0
suma_total_tienda3 = 1098019600.0
suma_total_tienda4 = 1038375700.0

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
totales = [suma_total_tienda1, suma_total_tienda2, suma_total_tienda3, suma_total_tienda4]

plt.figure(figsize=(12, 7))
plt.bar(tiendas, totales)

plt.title("Valor total de ventas por tienda")
plt.xlabel("Tienda")
plt.ylabel("Suma total (Precio)")
plt.grid(axis="y", linestyle="--", alpha=0.5)


for i, valor in enumerate(totales):
    plt.text(i, valor, f"{valor:,.0f}", ha='center', va='bottom')

plt.show()
```

# 2. Ventas por categoría

## Descripcion:
En este debes calcular la cantidad de productos vendidos por categoría en cada tienda. La idea es agrupar los datos por categoría y contar el número de ventas de cada tipo, mostrando las categorías más populares de cada tienda.


```python
ventas_tienda1 = tienda.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)
ventas_tienda2 = tienda2.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)
ventas_tienda3 = tienda3.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)
ventas_tienda4 = tienda4.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)
```

```python
print("TIENDA 1\n", ventas_tienda1, "\n")
print("TIENDA 2\n", ventas_tienda2, "\n")
print("TIENDA 3\n", ventas_tienda3, "\n")
print("TIENDA 4\n", ventas_tienda4, "\n")
```

#### Las categorias mas vendidas en todas las tiendas fueron **Muebles** y **Electronicos**.

## **Visualizacion Grafica**

```python
total_categorias = (
    ventas_tienda1
    .add(ventas_tienda2, fill_value=0)
    .add(ventas_tienda3, fill_value=0)
    .add(ventas_tienda4, fill_value=0)
)

total_categorias = total_categorias.sort_values(ascending=False)

plt.figure(figsize=(10, 10))
plt.pie(
    total_categorias.values,
    labels=total_categorias.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribución total de ventas por categoría (Tiendas 1, 2, 3 y 4)")
plt.axis("equal")
plt.show()
```

# 3. Calificación promedio de la tienda


## Descripcion:
En este paso, debes calcular las calificaciones promedio de los clientes para cada tienda. El objetivo es conocer la satisfacción del cliente con los productos vendidos.

```python
prom_tienda1 = tienda["Calificación"].mean()
prom_tienda2 = tienda2["Calificación"].mean()
prom_tienda3 = tienda3["Calificación"].mean()
prom_tienda4 = tienda4["Calificación"].mean()
```

```python
print("Calificación promedio - Tienda 1:", prom_tienda1)
print("Calificación promedio - Tienda 2:", prom_tienda2)
print("Calificación promedio - Tienda 3:", prom_tienda3)
print("Calificación promedio - Tienda 4:", prom_tienda4)
```

### La tienda con el mejor promedio de calificacion es, la **Tienda 3**, con un promedio de **4.05**

```python
prom_tienda1 = 3.976685036032217
prom_tienda2 = 4.037303942348453
prom_tienda3 = 4.048325561678677
prom_tienda4 = 3.9957591178965224

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
totales = [prom_tienda1, prom_tienda2, prom_tienda3, prom_tienda4]

plt.figure(figsize=(12, 7))
plt.bar(tiendas, totales)

plt.title("Calificación promedio de la tienda")
plt.xlabel("Tienda")
plt.ylabel("Promedio")
plt.grid(axis="y", linestyle="--", alpha=0.5)


for i, valor in enumerate(totales):
    plt.text(i, valor, f"{valor:,.0f}", ha='center', va='bottom')

plt.show()
```
